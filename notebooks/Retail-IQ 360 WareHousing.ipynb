 # dim_date 

```python
Retail_Main['order_date'] = pd.to_datetime(
    Retail_Main['order date (DateOrders)'], 
    errors='coerce'
)
```

```python
dim_date = pd.DataFrame({
    'full_date': Retail_Main['order_date'].drop_duplicates()
})
```

```python
dim_date['year'] = dim_date['full_date'].dt.year
dim_date['month'] = dim_date['full_date'].dt.month
dim_date['month_name'] = dim_date['full_date'].dt.month_name()
dim_date['quarter'] = dim_date['full_date'].dt.quarter
dim_date['day'] = dim_date['full_date'].dt.day
dim_date['weekday'] = dim_date['full_date'].dt.day_name()
dim_date['is_weekend'] = dim_date['weekday'].isin([
    'Saturday',
    'Sunday'
])
```

```python
dim_date.to_sql(
    'dim_date',
    conn,
    if_exists='replace',
    index=False
)
```

# from here i will add the col of shipping mode and the order status in the fact_order_items_3_final table from the retail_main table 
# so that i can make a relation between the dim_shipping table and hte fact_order_items_3_final table.


```python
fact_order_items_3_final=pd.read_sql('''select * from fact_order_items_3_final''',conn)
```

```python
Retail_main= pd.read_sql('''select * from RetailIQ360''',conn)
```

```python
# If fact_order_items_3_final already exists as a dataframe, just merge in the new columns
fact_order_items_3_finals = fact_order_items_3_final.merge(
    Retail_Main[['Order Id', 'Shipping Mode', 'Order Status']].drop_duplicates(),
    left_on='order_id',
    right_on='Order Id',
    how='left'
)

fact_order_items_3_finals = fact_order_items_3_finals.rename(columns={
    'Shipping Mode': 'shipping_mode',
    'Order Status': 'order_status'
}).drop(columns=['Order Id'])

print(fact_order_items_3_finals.columns.tolist())
print(fact_order_items_3_finals[['order_id','shipping_mode','order_status']].head())
```

```python
# Remove duplicate columns, keeping only the first occurrence of each name
fact_order_items_3_finals = fact_order_items_3_finals.loc[:, ~fact_order_items_3_finals.columns.duplicated()]

print(fact_order_items_3_finals.columns.tolist())
print(fact_order_items_3_finals.head())
```

```python
# Make sure order_date is datetime
fact_order_items_3_finals['order_date'] = pd.to_datetime(fact_order_items_3_finals['order_date'])

# Keep rows where year is NOT 2018
fact_order_items_3_finals = fact_order_items_3_finals[
    fact_order_items_3_finals['order_date'].dt.year != 2018
]
```

```python
fact_order_items_3_finals.to_sql(
    'fact_order_items_3_finals',
    conn,
    if_exists='replace',
    index=False
)
```
